```python
# -*- coding: utf-8 -*-
"""TLP Accessible with CodeCarbon Tracking

Automatically generated by Colab.
"""

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from codecarbon import EmissionsTracker, track_emissions

# Option 1: Tracking a specific function execution using the decorator
@track_emissions(project_name="tlp_plot_rendering")
def render_accessible_tlp(df_top, cvd_mode=True):
    """
    Renders TLP with CVD-friendly colors, texture hatching, and explicit icons.
    """
    fig, ax = plt.subplots(figsize=(8, 4))

    # Select Palette based on Accessibility Settings
    color_risk = '#D55E00' if cvd_mode else '#D9534F'   # Vermillion / Red
    color_safe = '#0072B2' if cvd_mode else '#5CB85C'   # Blue / Green

    bars = []
    for idx, row in df_top.iterrows():
        is_risk = row['shap_contribution'] > 0
        color = color_risk if is_risk else color_safe
        hatch = '///' if is_risk else '...'              # Redundant texture
        icon = "[▲] " if is_risk else "[▼] "             # Redundant symbol

        bar = ax.barh(icon + row['feature'], row['shap_contribution'],
                      color=color, hatch=hatch, edgecolor='black', linewidth=0.8)
        bars.append(bar)

    ax.axvline(0, color='black', linestyle='--', linewidth=1)
    ax.set_xlabel("SHAP Value Contribution (Risk Impact)")
    ax.set_title("Accessible Traffic Light Plot (CVD Mode)")
    plt.tight_layout()
    plt.show()


# Option 2: Wrap the entire execution script using CodeCarbon Context Manager
with EmissionsTracker(project_name="tlp_full_pipeline") as tracker:

    # --- Initial Sample Run ---
    data = {
        'feature': ['Feature A', 'Feature B', 'Feature C', 'Feature D', 'Feature E'],
        'shap_contribution': [0.5, -0.3, 0.7, -0.1, 0.4]
    }
    df_sample = pd.DataFrame(data)

    # Render the plot in CVD mode
    render_accessible_tlp(df_sample, cvd_mode=True)


    # --- Simulated pcI Dataset (Software Defects) ---
    dummy_pci_data = {
        'lines_of_code': np.random.randint(100, 5000, 5),
        'cyclomatic_complexity': np.random.randint(5, 50, 5),
        'developer_experience_months': np.random.randint(6, 120, 5),
        'test_coverage_percentage': np.random.randint(30, 99, 5),
        'number_of_commits': np.random.randint(10, 200, 5)
    }
    df_pci_software_defect = pd.DataFrame(dummy_pci_data)

    # Extract feature names
    realtime_feature_names_pci = df_pci_software_defect.columns.tolist()

    print("Simulated pcI dataset columns (software defect feature names):")
    print(realtime_feature_names_pci)

    new_shap_contributions_pci = [0.7, 0.5, -0.3, 0.6, -0.2]

    if len(realtime_feature_names_pci) != len(new_shap_contributions_pci):
        print(f"Warning: Feature count mismatch. Adjusting.")
        if len(realtime_feature_names_pci) > len(new_shap_contributions_pci):
            new_shap_contributions_pci.extend([0] * (len(realtime_feature_names_pci) - len(new_shap_contributions_pci)))
        else:
            new_shap_contributions_pci = new_shap_contributions_pci[:len(realtime_feature_names_pci)]

    df_sample_pci = pd.DataFrame({
        'feature': realtime_feature_names_pci,
        'shap_contribution': new_shap_contributions_pci
    })

    print("New df_sample with realtime pcI software defect feature names:")
    display(df_sample_pci)

    # Render complete pcI plot
    render_accessible_tlp(df_sample_pci, cvd_mode=True)


    # --- Top 5 Highest Risk Features ---
    df_risk_features = df_sample_pci[df_sample_pci['shap_contribution'] > 0]
    df_top_risk_features = df_risk_features.sort_values(by='shap_contribution', ascending=False).head(5)

    print("Top 5 Highest Risk Features:")
    display(df_top_risk_features)
    render_accessible_tlp(df_top_risk_features, cvd_mode=True)


    # --- Top 5 Lowest Risk Features ---
    df_low_risk_features = df_sample_pci[df_sample_pci['shap_contribution'] < 0]
    df_top_10_low_risk_features = df_low_risk_features.sort_values(by='shap_contribution', ascending=True).head(5)

    print(f"Displaying {len(df_top_10_low_risk_features)} lowest risk features:")
    display(df_top_10_low_risk_features)
    render_accessible_tlp(df_top_10_low_risk_features, cvd_mode=True)


    # --- All Risk Features ---
    df_all_risk_features = df_sample_pci[df_sample_pci['shap_contribution'] > 0]
    print(f"Displaying {len(df_all_risk_features)} all risk features:")
    display(df_all_risk_features)
    render_accessible_tlp(df_all_risk_features, cvd_mode=True)


    # --- All Low Risk Features ---
    df_all_low_risk_features = df_sample_pci[df_sample_pci['shap_contribution'] < 0]
    print(f"Displaying {len(df_all_low_risk_features)} all low risk features:")
    display(df_all_low_risk_features)
    render_accessible_tlp(df_all_low_risk_features, cvd_mode=True)

# Inspection code: View the saved emissions log
emissions_df = pd.read_csv("emissions.csv")
print("\n--- CODECARBON EMISSIONS SUMMARY ---")
display(emissions_df[["project_name", "duration", "emissions", "cpu_energy", "gpu_energy"]])
```

```python
mean_emissions = emissions_df['emissions'].mean()
print(f"Mean emissions across all runs: {mean_emissions:.6f} kg CO2eq")
```

```python
# Calculate average carbon intensity per project
# Carbon intensity = total emissions / total duration for each project

project_avg_emissions_duration = emissions_df.groupby('project_name')[['emissions', 'duration']].mean().reset_index()
project_avg_emissions_duration['carbon_intensity_kg_co2eq_per_second'] = project_avg_emissions_duration['emissions'] / project_avg_emissions_duration['duration']

print("\n--- Average Carbon Intensity per Project (kg CO2eq/second) ---")
display(project_avg_emissions_duration[['project_name', 'carbon_intensity_kg_co2eq_per_second']])
```

```python
# Calculate total energy consumption by project
project_energy_consumption = emissions_df.groupby('project_name')[['cpu_energy', 'gpu_energy', 'ram_energy']].sum().reset_index()

print("\n--- Total Energy Consumption by Project (kWh) ---")
display(project_energy_consumption)
```

```python
project_energy_consumption['total_energy'] = project_energy_consumption['cpu_energy'] + project_energy_consumption['gpu_energy'] + project_energy_consumption['ram_energy']
project_energy_consumption['ram_percentage'] = (project_energy_consumption['ram_energy'] / project_energy_consumption['total_energy']) * 100

print("\n--- RAM Energy Percentage by Project ---")
display(project_energy_consumption[['project_name', 'ram_percentage']])
```

### Comparison of CodeCarbon Results: 'SHAP and TLP Pipeline' vs. 'TLP Plot Rendering'

Based on the `emissions_df`, we can compare the following two projects:

*   **`tlp_full_pipeline`**: This project tracks the emissions for the entire script execution, including the generation of dummy SHAP-like data (`df_sample_pci`) and all subsequent `render_accessible_tlp` function calls within its context manager.
*   **`tlp_plot_rendering`**: This project specifically tracks the emissions generated by each individual call to the `render_accessible_tlp` function, thanks to the `@track_emissions` decorator.

Let's summarize their key metrics:


```python
# Display total emissions for comparison
print("\n--- Total Emissions Comparison (kg CO2eq) ---")
display(project_emissions)

# Display average carbon intensity for comparison
print("\n--- Average Carbon Intensity Comparison (kg CO2eq/second) ---")
display(project_avg_emissions_duration[['project_name', 'carbon_intensity_kg_co2eq_per_second']])

# Display energy consumption breakdown for comparison
print("\n--- Energy Consumption Breakdown Comparison (kWh) ---")
display(project_energy_consumption[['project_name', 'cpu_energy', 'gpu_energy', 'ram_energy', 'total_energy']])
```

#### Observations:

*   **Total Emissions**: The `tlp_full_pipeline` (which includes all steps from data generation to multiple TLP renderings) has a slightly higher total emission than `tlp_plot_rendering` (which aggregates only the emissions from the TLP function calls).
*   **Carbon Intensity**: Both projects show very similar and extremely low carbon intensities (emissions per second), indicating that the computational work involved is very light in terms of CO2 equivalent per unit of time.
*   **Energy Consumption**: Similar to emissions, `tlp_full_pipeline` consumed marginally more CPU and RAM energy, which is expected as it encompasses a broader scope of operations. GPU consumption remains zero for both as no GPU was detected or utilized.
